```python
# import packages
import torch
import torch.nn as nn # imports neural network module
import torch.optim as optim # imports optimizer

class Autoencoder(nn.Module): # autoencoders use the nn module
    def __init__(self): 
        super().__init__()
        self.encoder=nn.Linear(8,16) # encoder/linear layer 1
        self.decoder=nn.Linear(16,8) # decoder/linear layer 2

    def forward(self,X): # what the autoencoder does when it is called
        encoded = self.encoder(X) # latent representations 
        x=self.decoder(encoded)
        return x

X =  torch.tensor([
    [0.2, 0.5, 0.1, 0.8, 0.3, 0.0, 0.4, 0.7],
    [0.9, 0.1, 0.6, 0.2, 0.5, 0.3, 0.8, 0.4],
    [0.1, 0.7, 0.2, 0.9, 0.4, 0.2, 0.3, 0.6],
    [0.8, 0.2, 0.5, 0.1, 0.7, 0.4, 0.9, 0.3]
])

model = Autoencoder() # creates a new model
criterion =nn.MSELoss() # mean square error : mean squared difference between prediction values and target valuse
optimizer= torch.optim.SGD(model.parameters(), lr = 0.01) # using SGD optimizer to move weights in the opposite direction of the directions calculated by the gradient , to decrease the loss

for epoch in range(1000): # takes all input at once 1000 times
    reconstruction = model(X) # calls the forward function
    reconstruction_loss= criterion(reconstruction, X) # compares the reconstruction of input with the orignal input

    optimizer.zero_grad() # clears gradients from the previous training sets
    reconstruction_loss.backward() # calculates the gradients
    optimizer.step() # updates the weights

    if (epoch+1) % 100 == 0:
        print(f"Epoch : {epoch + 1}, Loss : {reconstruction_loss.item():.6f} ") # prints the epoch and its respective loss



```
