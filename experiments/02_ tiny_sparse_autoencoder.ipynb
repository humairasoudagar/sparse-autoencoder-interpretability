```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
class SparseAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Linear(8, 16)
        self.decoder = nn.Linear(16, 8)

    def forward(self, X):
        encoded = self.encoder(X)
        features = torch.relu(encoded)
        reconstruction = self.decoder(features)

        return reconstruction, features
```

```python
X = torch.tensor([
    [0.2, 0.5, 0.1, 0.8, 0.3, 0.0, 0.4, 0.7],
    [0.9, 0.1, 0.6, 0.2, 0.5, 0.3, 0.8, 0.4],
    [0.1, 0.7, 0.2, 0.9, 0.4, 0.2, 0.3, 0.6],
    [0.8, 0.2, 0.5, 0.1, 0.7, 0.4, 0.9, 0.3]
])




    
```

```python
model = SparseAutoencoder()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(),lr=0.01)

lambda_ = 0.01
```

```python
reconstruction, features = model(X)

print("Reconstruction:")
print(reconstruction)

print("\nFeatures:")
print(features)
```

```python
for epoch in range(1000):

    reconstruction, features = model(X)
    reconstruction_loss = criterion(reconstruction, X)
    sparse_penalty = torch.sum(torch.abs(features))

    loss = reconstruction_loss + lambda_ * sparse_penalty

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 100 == 0:
        print(f"Epoch: {epoch + 1}, Loss: {loss.item():.6f}")
```
